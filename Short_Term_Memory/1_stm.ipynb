```python
from langgraph.graph import StateGraph, START, MessagesState
# from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# model = ChatOpenAI()
from langchain_ollama import ChatOllama

# Initialize the model
model = ChatOllama(
    model="granite4.1:3b"
)
```

```python
def call_model(state: MessagesState):
    response = model.invoke(state["messages"])
    return {"messages": [response]}
```

```python
builder = StateGraph(MessagesState)
builder.add_node("call_model", call_model)
builder.add_edge(START, "call_model")
```

```python
graph = builder.compile()
graph
```

```python
graph.invoke({"messages": [{"role": "user", "content": "Hi! I am a Student and learning LangGraph."}]})
```

```python
graph.invoke({"messages": [{"role": "user", "content": "What am I learning?"}]})
```

```python
def call_model(state: MessagesState):
    response = model.invoke(state["messages"])
    return {"messages": [response]}
```

```python
builder = StateGraph(MessagesState)
builder.add_node("call_model", call_model)
builder.add_edge(START, "call_model")
```

```python
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()
graph = builder.compile(checkpointer=checkpointer)
```

```python
config = {"configurable": {"thread_id": "thread-1"}}
config2 = {"configurable": {"thread_id": "thread-2"}}
```

```python
graph.invoke({"messages": [{"role": "user", "content": "Hi! I am a Student and learning LangGraph."}]}, config)
```

```python
graph.invoke({"messages": [{"role": "user", "content": "What am I learning?"}]}, config2)
```

```python
snap = graph.get_state(config)
vals = snap.values
for m in vals.get("messages", []):
        print("-", type(m).__name__, ":", m.content)
```
